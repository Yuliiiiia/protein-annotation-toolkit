# Protein Structure Viewer

Interactive 3D visualization + annotation report for any PDB entry.

Requires: `pip install py3Dmol` (in addition to the toolkit requirements).

```python
import sys
sys.path.insert(0, "..")

import py3Dmol
from protein_toolkit.fetch import fetch_structure
from protein_toolkit.analyze import analyze_structure
from protein_toolkit.report import render_markdown
from IPython.display import Markdown, display
```

```python
PDB_ID = "1CRN"  # try: 1CRN (crambin, small+fast), 6VXX (SARS-CoV-2 spike), 1UBQ (ubiquitin)

structure_path = fetch_structure(PDB_ID, out_dir="../data")
print(f"Downloaded to {structure_path}")
```

## 3D Structure

```python
with open(structure_path) as f:
    structure_data = f.read()

view = py3Dmol.view(width=700, height=500)
view.addModel(structure_data, "cif")
view.setStyle({"cartoon": {"color": "spectrum"}})
view.addStyle({"hetflag": True}, {"stick": {"colorscheme": "greenCarbon"}})
view.zoomTo()
view.show()
```

## Annotation Report

```python
results = analyze_structure(structure_path)
display(Markdown(render_markdown(results)))
```

## Color by secondary structure (if DSSP is installed)

```python
view2 = py3Dmol.view(width=700, height=500)
view2.addModel(structure_data, "cif")
view2.setStyle({"cartoon": {"colorscheme": "amino"}})
view2.zoomTo()
view2.show()
```
